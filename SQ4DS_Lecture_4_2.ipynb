# 1. What is Machine Learning?

A useful beginner definition is:

> **Machine Learning is a way of building computer systems that learn useful patterns from data so they can make predictions, classifications, recommendations, or decisions.**

The central idea is that we do not necessarily write every rule by hand. Instead, we provide examples, an objective, and a learning procedure, and the model learns parameters that help it perform the task.

### A simple comparison

**Traditional programming**

`Rules + Data → Program → Output`

**Machine Learning**

`Examples + Learning algorithm → Model`

Then:

`New data + Trained model → Prediction`

The programmer still decides what problem to solve, what data to use, what objective matters, and how the model is evaluated. The difference is that part of the decision logic is **learned from examples**.

# 2. Why was Machine Learning needed?

Traditional programming is excellent when a problem can be described with explicit rules.

For example:

> If an order is above ₦50,000, apply free delivery.

That rule is clear. We can simply program it.

But many real-world tasks are difficult to express as a manageable list of rules:

- Is this image a cat or a dog?
- Is this transaction suspicious?
- What word is likely to come next?
- Which products may interest this customer?
- How much demand should we expect next week?
- Does this email look like spam?

For problems like these, there may be **too many interacting patterns** for humans to specify manually. Learning from examples can be more practical.

### The main idea

Machine Learning becomes useful when the relationship between inputs and desired outputs is difficult to program explicitly but there is enough representative data from which a useful pattern can be learned.

# 3. When should you NOT use Machine Learning?

This is an important part of becoming a good practitioner.

You do not need ML simply because a problem contains data.

A traditional program, SQL query, spreadsheet calculation, statistical analysis, or business rule may be better when:

- the rules are clear and stable
- the logic is deterministic
- the problem is simple
- there is little suitable training data
- model maintenance would add unnecessary complexity
- the consequences of errors require a different type of controlled system
- a simpler solution produces the same practical result

### Principle

> **Use Machine Learning when learning from data adds value—not merely because ML is available.**

# 4. A brief history of Machine Learning

Machine Learning grew out of several fields, including statistics, artificial intelligence, optimization, pattern recognition, and computer science.

### 1940s–1950s: foundations

Early researchers investigated whether machines could imitate aspects of human reasoning. Alan Turing's 1950 work on machine intelligence became an influential landmark in the history of AI.

### 1950s–1960s: early learning systems

Researchers began creating systems that adjusted themselves using examples. The **perceptron**, associated with Frank Rosenblatt in the late 1950s, became an important early neural-network model.

### 1970s–1980s: expert systems and learning research

AI research included systems based on explicitly encoded expert knowledge, while statistical and neural learning approaches continued to develop. Backpropagation later became central to training multi-layer neural networks.

### 1990s: statistical Machine Learning expands

Methods such as decision trees, support vector machines, Bayesian methods, and ensemble approaches became widely studied. The field increasingly emphasized learning from data rather than encoding every rule manually.

### 2000s: data and computing scale up

The growth of the internet created large datasets. Faster processors, GPUs, distributed computing, and improved optimization made larger models practical.

### 2010s: Deep Learning accelerates

Deep neural networks made major advances in image recognition, speech, and language-related tasks, supported by larger datasets and powerful hardware.

### 2020s: foundation models and Generative AI

Large models became capable of generating and transforming text, images, audio, code, and other types of content.

### The recurring pattern

The history of ML repeatedly shows that progress comes from combinations of:

**better algorithms + better data + more computing + better engineering**

# 6. Traditional Software Engineering vs Machine Learning

The most useful distinction is **where the behavior of the system comes from**.

## Traditional software

A developer explicitly writes rules and logic.

```text
Inputs → Hand-written logic → Outputs
```

## Machine Learning

The developer defines a problem and learning process, while model parameters are learned from examples.

```text
Training examples → Learning process → Learned model

New input → Learned model → Prediction
```

### But ML is still software engineering

A production ML system still needs:

- requirements
- architecture
- testing
- version control
- deployment
- security
- documentation
- monitoring
- maintenance

The difference is that an ML system has **data and learned behavior as first-class engineering components**.

## Side-by-side comparison

| Aspect | Traditional software | Machine Learning |
|---|---|---|
| Main behavior | Explicit rules | Learned patterns |
| Main ingredients | Code + inputs | Data + objective + algorithm + code |
| Testing | Check expected outputs/behavior | Check performance on unseen data |
| Typical update | Change code/rules | Often retrain/update model and/or data |
| Main new risk | Implementation bugs | Data/model/implementation failures |
| Monitoring | System and application health | System + data + model behavior |
| Explainability | Often easier to trace | Depends on model and explanation method |
| Maintenance | Mostly code/infrastructure | Code + data + model + infrastructure |

# 7. What makes an ML problem different?

In traditional software, you can often ask:

> "Did the program implement the rules correctly?"

For ML, we also need to ask:

> "Did the model learn a useful relationship that generalizes to new cases?"

This introduces concepts that are central to ML:

- training data
- validation and test data
- generalization
- loss functions
- statistical uncertainty
- overfitting
- data leakage
- distribution shift

That is why ML needs a different style of testing and evaluation.

# 8. Anatomy of a Machine Learning problem

Nearly every beginner ML problem can be described with a few terms.

### Observation / Sample

One example or row in the dataset.

### Feature (X)

An input variable used by the model.

Examples:

- age
- income
- temperature
- number of transactions
- word counts
- pixel values

### Target / Label (y)

The outcome we want to predict in supervised learning.

Examples:

- price
- spam/not spam
- churn/not churn
- disease class

### Model

A function or computational system that maps inputs to outputs.

### Parameter

A value learned from training data by the model.

### Hyperparameter

A configuration selected outside the model's ordinary parameter-learning process.

# 9. A simple example: house-price prediction

Suppose each row represents one house.

| size | bedrooms | location_score | price |
|---:|---:|---:|---:|
| 1200 | 2 | 6.5 | 25,000,000 |
| 1500 | 3 | 7.5 | 34,000,000 |
| 2000 | 4 | 8.2 | 48,000,000 |

Here:

- **Features:** size, bedrooms, location_score
- **Target:** price
- **Task:** regression
- **Training:** learn a relationship between features and price
- **Inference:** predict the price of a new house

The same structure appears in many ML problems—even when the domain changes completely.

# 10. The five questions behind most ML problems

Before coding, ask:

### 1. What is one row?
What exactly does one observation represent?

### 2. What are the inputs?
Which information is available to the model?

### 3. What is the output?
What exactly are we trying to predict?

### 4. When must the prediction be made?
What information is available **at that moment**?

### 5. What counts as a useful prediction?
Which errors matter, and how will success be measured?

These questions prevent many common beginner mistakes before a model is trained.

# 11. The main types of Machine Learning

A useful first classification is:

1. **Supervised Learning** — learn from labeled examples
2. **Unsupervised Learning** — discover structure without a target
3. **Semi-supervised Learning** — combine labeled and unlabeled examples
4. **Self-supervised Learning** — create learning signals from the data itself
5. **Reinforcement Learning** — learn through actions and rewards

The first two are the most important to master in the early stages of practical Data Science.

# 12. Supervised Learning

In supervised learning we have inputs `X` and a known target `y`.

The model learns a relationship:

$$f(X) \approx y$$

The two major beginner tasks are:

### Classification
Predict a category.

Examples:

- fraud / not fraud
- spam / not spam
- product category

### Regression
Predict a numeric value.

Examples:

- house price
- sales amount
- energy demand

# 13. Classification in more detail

### Binary classification

Two classes.

Examples: yes/no, 0/1, fraud/not fraud.

### Multiclass classification

More than two possible classes.

Examples: A/B/C or several species.

### Multilabel classification

One observation can have several labels at once.

Example: a document can belong to both **technology** and **business**.

A classification model may produce:

- a predicted class
- class probabilities
- a score

The useful choice depends on the application.

# 14. Regression in more detail

Regression predicts a continuous numerical outcome.

A simple linear model is:

$$\hat{y} = w_0 + w_1x_1 + w_2x_2 + \cdots + w_px_p$$

The values `w` are learned from the training data.

Linear regression is important even when you later use more advanced models because it provides a simple, interpretable baseline.

# 15. Unsupervised Learning

In unsupervised learning we have features `X` but no target `y` supplied for the task.

We may want to discover structure such as:

### Clustering
Group similar observations.

Example: customer segmentation.

### Dimensionality reduction
Represent many features using fewer dimensions.

Example: reducing 100 variables to 2 for visualization.

### Anomaly detection
Identify observations that look unusual.

Example: unusual transaction patterns.

A challenge with unsupervised learning is that there may be no single correct answer. Interpretation becomes especially important.

# 16. Semi-supervised and self-supervised learning

### Semi-supervised learning

Imagine 1 million examples exist but only 10,000 are labeled. Semi-supervised methods can use both the labeled and unlabeled data.

### Self-supervised learning

The data itself is used to create the learning task.

For example, a system can hide part of an input and learn to predict the hidden portion.

Self-supervised learning has become an important foundation for many modern large-scale models.

# 17. Reinforcement Learning

Reinforcement Learning is built around **sequential decisions**.

A simplified setup contains:

- **agent** — the learner
- **environment** — the world it interacts with
- **state** — current situation
- **action** — what the agent does
- **reward** — feedback from the environment

The goal is usually to learn a strategy that maximizes long-term reward.

Reinforcement Learning is common in areas involving repeated decisions, such as robotics, games, and some control problems.

# 19. The ML learning cycle

A model does not magically become intelligent after receiving data.

A simplified learning cycle is:

```text
Training data
     ↓
Preprocessing
     ↓
Model
     ↓
Prediction
     ↓
Loss / error
     ↓
Parameter update
     ↓
Repeat
```

The exact process depends on the algorithm, but the central idea is that the model adjusts its parameters to improve an objective.

# 20. Loss functions: how does a model know it is wrong?

A learning algorithm needs a way to quantify error.

This is often called a **loss function** or **objective function**.

### Regression example

Mean Squared Error (MSE):

$$MSE = \frac{1}{n}\sum_{i=1}^{n}(y_i-\hat{y}_i)^2$$

### Classification

Probabilistic classifiers often use losses such as **log loss / cross-entropy**.

The learning process tries to find parameters that make the objective smaller, subject to the chosen algorithm and optimization method.

# 21. Training vs inference

### Training

The model uses historical examples to learn parameters.

### Inference

The trained model receives new information and produces an output.

Example:

**Training:** historical transactions → learn fraud patterns

**Inference:** new transaction → produce a fraud probability

Inference can happen:

- in real time
- in batches
- on a server
- on a mobile/edge device

A model's behavior in production depends on both the model and the system around it.

# 22. Generalization: the real goal

A useful ML model should work on **unseen data**, not merely reproduce the training examples.

This ability is called **generalization**.

A model can perform extremely well on training data and still fail on new observations.

That is why ML evaluation is centered on data the model did not use to learn its parameters.

# 23. Training, validation, and test data

A common setup is:

### Training set
Used to learn model parameters.

### Validation set
Used while developing the model to compare choices and tune hyperparameters.

### Test set
Held back for final evaluation.

For many smaller datasets, cross-validation is used during model selection.

### Golden rule

> **Do not repeatedly optimize your choices using the final test set.**

Otherwise, the test set is no longer a trustworthy measure of unseen performance.

# 24. Data leakage

**Data leakage** occurs when information enters training or evaluation that would not legitimately be available when the prediction is made.

Common examples:

### Target leakage
A feature contains information created after the target event.

### Preprocessing leakage
A scaler or imputer is fitted using information from the test set.

### Duplicate leakage
The same entity or near-duplicate observations appear across train and test when they should not.

### Time leakage
Future information is used to predict the past.

Leakage can produce extremely impressive—but misleading—scores.

# 25. Overfitting and underfitting

### Underfitting

The model is too simple to capture important structure.

Typical pattern:

**poor training performance + poor test performance**

### Overfitting

The model captures training-specific detail that does not generalize.

Typical pattern:

**very good training performance + substantially worse test performance**

### The goal

Find enough complexity to learn useful patterns without becoming too dependent on the training sample.

# 26. Bias and variance

A useful conceptual model is:

### High bias

The model makes strong simplifying assumptions and misses important structure.

### High variance

The model is very sensitive to the exact training sample.

This is often described as the **bias-variance trade-off**.

Practical tools such as regularization, feature selection, tree constraints, more data, and cross-validation can help control this trade-off.

# 29. Data preparation and preprocessing

Real datasets often contain:

- missing values
- inconsistent categories
- outliers
- wrong types
- mixed units
- duplicated records
- highly skewed variables

Typical preprocessing includes:

- imputation
- encoding categorical data
- scaling numerical features
- transformations
- feature selection

### Important

Preprocessing is part of the learned-system workflow. It must be applied consistently to training data and new data, without leaking information from the evaluation set.

# 30. Feature engineering

Feature engineering means creating or transforming inputs so that useful information is easier for the model to use.

Examples:

- month from a date
- total spending in the previous 30 days
- debt-to-income ratio
- number of support calls
- interaction between two measurements
- logarithm of a heavily skewed variable

Good feature engineering combines **technical skill and domain knowledge**.

A feature can be highly predictive but still invalid if it uses information unavailable at prediction time.

# 31. Feature scaling

Some algorithms are sensitive to the numerical scale of features.

For example:

- KNN
- K-Means
- Support Vector Machines
- many gradient-based models

A common standardization is:

$$z=\frac{x-\mu}{\sigma}$$

Tree-based methods generally do not require scaling.

# 32. Categorical data

Categorical values such as:

`city = Lagos, Abuja, Kano`

usually need to be represented numerically before being used by many ML algorithms.

A common approach is **one-hot encoding**.

Other approaches may be appropriate for high-cardinality or ordered categories.

The correct encoding depends on the meaning of the variable and the model.

# 33. Pipelines

A pipeline combines transformations and a model into a single repeatable workflow.

Conceptually:

```text
Raw data
   ↓
Imputation
   ↓
Encoding / scaling
   ↓
Model
   ↓
Prediction
```

Pipelines help:

- reduce accidental leakage
- keep transformations consistent
- make cross-validation safer
- make experiments reproducible
- simplify production workflows

# 34. Cross-validation

Cross-validation helps us estimate how a model is likely to perform across different samples of the training data.

In k-fold cross-validation:

1. divide the training data into k parts
2. train on k−1 parts
3. validate on the remaining part
4. repeat until every part has served as validation data
5. summarize the results

Cross-validation is mainly a **development/model-selection tool**. It does not remove the need for a final test evaluation when a protected test set is appropriate.

# 35. Hyperparameters

A **parameter** is learned by the model from data.

A **hyperparameter** is a configuration chosen before or around training.

Examples:

- tree depth
- number of trees
- number of neighbors in KNN
- regularization strength
- learning rate
- number of hidden layers

Hyperparameters can be selected using domain knowledge, grid search, random search, or other optimization strategies.

# 36. Classification metrics

### Accuracy

Fraction of predictions that are correct.

Useful when class distribution and error costs make it appropriate.

### Precision

Of the predicted positive cases, how many are actually positive?

### Recall

Of the actual positive cases, how many did we detect?

### F1 score

A harmonic mean of precision and recall.

### Confusion matrix

Shows the counts of different types of correct and incorrect classification decisions.

### Other useful tools

Depending on the problem, you may also use:

- ROC-AUC
- precision-recall curves / PR-AUC
- log loss
- calibration measures

The metric should match the real-world objective.

# 37. Regression metrics

### MAE — Mean Absolute Error

Average absolute difference between actual and predicted values.

Easy to interpret in the target's units.

### MSE — Mean Squared Error

Squares errors, so larger errors receive more weight.

### RMSE — Root Mean Squared Error

Square root of MSE, returning to the target's original units.

### R²

Describes performance relative to a mean-based baseline under standard regression assumptions.

No single metric tells the entire story. Understand the business/research meaning of the error.

# 38. Class imbalance

Imagine only 0.5% of transactions are fraudulent.

A system that predicts **"not fraud" for every transaction** could have very high accuracy while being useless for fraud detection.

In imbalanced problems, inspect metrics such as:

- precision
- recall
- F1
- PR-AUC
- ROC-AUC
- confusion matrix

Also think about the **cost of each type of error**.

# 39. Probability and decision thresholds

A classifier may output a probability such as:

> `fraud probability = 0.83`

A threshold converts that score into an action.

For example:

```text
probability >= 0.50 → flag
probability < 0.50  → do not flag
```

But 0.50 is not a universal rule. A real application might use a different threshold depending on the consequences of false positives and false negatives.

Therefore:

**model scoring** and **decision policy** are related but distinct parts of an ML system.

# 40. Error analysis

After evaluating a model, do not stop at one score.

Ask:

- Which examples did the model get wrong?
- Are errors concentrated in one class?
- Are some groups or time periods affected more than others?
- Are errors caused by missing features?
- Are labels themselves noisy?
- Is there a systematic pattern?

Error analysis often produces more useful improvement ideas than simply trying another algorithm.

# 41. Correlation is not causation

Machine Learning is frequently used for prediction.

Prediction and causal inference are different goals.

If a feature is strongly associated with the target, that does **not** automatically mean changing the feature will change the target.

Possible reasons include:

- confounding variables
- reverse causality
- selection effects
- measurement issues

Always distinguish **predictive usefulness** from **causal explanation**.

# 42. Interpretability

### Tree ensembles

### Important limitation

An explanation of model behavior is not automatically a proof of causation.

# 46. Model monitoring

The world can change while your code stays the same.

### Data drift

The distribution of input data changes.

### Concept drift

The relationship between inputs and the target changes.

### Performance degradation

The model's predictive quality declines.

### Operational degradation

Latency, failures, memory use, or service availability worsens.

A model is therefore not simply "built once." Production ML is a lifecycle of **build → deploy → monitor → improve**.

# 49. Minimal code: classification workflow

Now that the concepts are in place, the code becomes easier to understand.

````python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

iris = load_iris(as_frame=True)

X = iris.data
Y = iris.target

X_train, X_test, y_train, y_test = train_test_split(
    X, Y,
    test_size=0.20,
    random_state=42,
    stratify=Y
)

model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)

pred = model.predict(X_test)

print('Accuracy:', accuracy_score(y_test, pred))
print('\nClassification report:')
print(classification_report(y_test, pred, target_names=iris.target_names))
````

## Connect the code to the theory

- `X` → features
- `Y` → target
- `train_test_split()` → separates seen and unseen data
- `fit()` → training / parameter learning
- `predict()` → inference
- `accuracy_score()` → evaluation

The code is intentionally short. The conceptual work—problem framing, data understanding, validation design, metric selection, and error analysis—is the bigger skill.

````python
cm = confusion_matrix(y_test, pred)
print('Confusion matrix:')
print(cm)
````

# 50. Minimal code: regression workflow

````python
from sklearn.datasets import load_diabetes
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Load data
diabetes = load_diabetes(as_frame=True)
X = diabetes.data
Y = diabetes.target

X_train, X_test, y_train, y_test = train_test_split(
    X, Y,
    test_size=0.20,
    random_state=42
)

model = LinearRegression()
model.fit(X_train, y_train)

pred = model.predict(X_test)

mae = mean_absolute_error(y_test, pred)
rmse = mean_squared_error(y_test, pred) ** 0.5
r2 = r2_score(y_test, pred)

print('MAE :', mae)
print('RMSE:', rmse)
print('R²  :', r2)
````

# 55. Final mental model

When you meet a new dataset, do **not** begin with:

> "Which algorithm should I use?"

Begin with:

### Problem
What are we actually trying to accomplish?

### Data
What evidence do we have, and how was it collected?

### Time
What information is available when the prediction must be made?

### Target
What exactly are we trying to predict?

### Baseline
What can a simple solution achieve?

### Model
Which family fits the problem and constraints?

### Evaluation
How will we measure success?

### Errors
Which mistakes matter most?

### Deployment
How will the prediction be used?

### Monitoring
How will we know when the model stops working well?

> **The real skill in Machine Learning is not memorizing algorithms. It is learning how to turn a real problem into a sound data, modeling, evaluation, and deployment process.**
