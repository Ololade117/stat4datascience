## Statistics is the process of using data to understand the world around us. The stages of statistics, often called the statistical investigation or analysis cycle, consist of five core steps: collection, organization, presentation, analysis, and interpretation






# The Five Stages of Statistics

Statistics  typically follow a systematic approach, which can be broken down into the following five stages:

1.  **Data Collection:**
    *   Collect raw, primary, or secondary data using surveys, experiments, interviews, or observations. Focus: **Ensure the sample is accurate and unbiased to properly represent the target population**

2.  **Data Organization:**
    *   Cleaning data: Review raw data to fix errors, remove inconsistencies, and organize information into rows and columns or arrays.Focus: **Make raw data ready for sorting, analysis and classification.**

3. **Data Presentation:**
    *    Display processed data using tables, charts, graphs, or diagrams like histograms or bar graphs. Focus: **Make complex numbers(no pun intended) easy to read and understand at a glance.**

4. **Data Analysis:**
    *    Apply mathematical and statistical formulas—such as averages, standard deviation, regression, or hypothesis testing. Focus: ** Extract meaningful patterns, trends, and relationships from the numbers.**

5. **Data Interpretation:**
    *    Explain what the analyzed data actually means in the context of your original research question.Focus: **Formulate final decisions, predictions, or generalizations about the larger population.**

## What is Data Science?

Data Science is an interdisciplinary field that uses scientific methods, processes, algorithms, and systems to extract knowledge and insights from structured and unstructured data. It combines elements of mathematics, statistics, computer science, and domain-specific knowledge to analyze data and make informed decisions.



# The Seven Stages of Data Science

Data science projects typically follow a systematic approach, which can be broken down into the following seven stages:

1.  **Problem Definition/Business Understanding:**
    *   Clearly identify the problem or question you want to answer. Understand the business objectives, constraints, and requirements. This stage ensures that the data science effort is aligned with organizational goals.

2.  **Data Acquisition/Collection:**
    *   Gather relevant data from various sources (databases, APIs, web scraping, logs, etc.). This involves identifying what data is needed and how to obtain it.

3.  **Data Cleaning/Preparation:**
    *   Clean, transform, and format the raw data. This often involves handling missing values, dealing with outliers, correcting inconsistencies, converting data types, and ensuring the data is in a suitable format for analysis. This is often the most time-consuming stage.

4.  **Data Exploration/Exploratory Data Analysis (EDA):**
    *   Analyze the characteristics of the data, discover patterns, detect anomalies, and test hypotheses using statistical graphics and other data visualization methods. EDA helps in understanding the data's structure and uncovering initial insights.

5.  **Feature Engineering:**
    *   Create new features or transform existing ones to improve the performance of machine learning models. This might involve combining features, extracting information from text, or scaling numerical data.

6.  **Model Building/Training:**
    *   Select appropriate machine learning algorithms and build predictive models. This includes splitting data into training and testing sets, training the model, and tuning its parameters to achieve optimal performance.

7.  **Model Evaluation & Deployment:**
    *   Assess the model's performance using relevant metrics (e.g., accuracy, precision, recall, F1-score, RMSE). Once a satisfactory model is developed, it is deployed into a production environment where it can be used to make predictions or inform decisions. This stage also includes monitoring the model's performance over time and iterating if necessary.


## How Statistics Works in Data Science

Statistics plays a foundational and integral role throughout the entire data science lifecycle. It provides the tools and methodologies necessary to understand, analyze, and interpret data, ultimately enabling data scientists to extract meaningful insights and make informed decisions.

Here's how statistics is applied across the stages of data science:

1.  **Problem Definition/Business Understanding:** Statistical thinking helps in framing the business problem into a quantifiable question, identifying the right metrics, and understanding the potential biases in data that might influence the solution.

2.  **Data Acquisition/Collection:** While data acquisition focuses on gathering data, statistical principles like sampling are crucial to ensure the collected data is representative of the target population and free from selection bias.

3.  **Data Cleaning/Preparation:** Descriptive statistics (mean, median, mode, standard deviation, quartiles) are used extensively to identify anomalies, outliers, and missing values. Statistical methods can guide decisions on imputation strategies (e.g., replacing missing values with the mean or median) and data transformation (e.g., normalization or standardization).

4.  **Data Exploration/Exploratory Data Analysis (EDA):** This stage is heavily reliant on statistics. Techniques like correlation analysis, hypothesis testing, variance analysis, and various statistical plots (histograms, box plots, scatter plots) are used to:
    *   Summarize the main characteristics of the data.
    *   Discover patterns and relationships between variables.
    *   Detect outliers and anomalies.
    *   Test assumptions and formulate hypotheses for further analysis.

5.  **Feature Engineering:** Statistical measures can help in measuring how much a feature correlates with other features to infer the significance of the feature

6.  **Model Building/Training:** Machine learning models are often rooted in statistical concepts. Regression models, classification algorithms (like logistic regression, SVMs), and even complex neural networks use statistical optimization techniques. Statistical inference is used to understand the significance of model parameters and the confidence in their estimates.

7.  **Model Evaluation & Deployment:** Statistical metrics are paramount for evaluating model performance. Metrics like accuracy, precision, recall, F1-score, AUC-ROC (for classification), RMSE, MAE (for regression) are all statistical measures. Hypothesis testing can be used to compare different models or to determine if a model's performance on new data is significantly different from its performance on training data. Statistical process control can also be used to monitor model performance post-deployment.

In essence, statistics provides the theoretical backbone and practical tools that empower data scientists to make sense of data, build robust models, and draw reliable conclusions, ensuring that insights are not just observed but are statistically significant and actionable.

## What is Data?

**Data** refers to raw facts, figures, observations, or information that is collected, stored, and processed. In the context of data science, data is the fundamental building block from which insights, knowledge, and decisions are derived. It can exist in many forms, from numbers and text to images and sounds.

## Types of Data

Data can be broadly categorized in several ways, often based on its nature and the type of measurements it represents. The two main overarching categories are **Qualitative** and **Quantitative** data.

### 1. Qualitative Data (Categorical Data)

Qualitative data describes qualities or characteristics that cannot be measured numerically. It often falls into categories and can be observed but not directly calculated. This type of data helps to understand 'why' or 'how' things are.

*   **Nominal Data:** This data consists of categories that have no natural order or ranking. Examples include:
    *   **Gender:** Male, Female, Non-binary
    *   **Marital Status:** Single, Married, Divorced, Widowed
    *   **Eye Color:** Blue, Brown, Green, Hazel
    *   **Favorite Color:** Red, Blue, Green

*   **Ordinal Data:** This data consists of categories that have a natural order or ranking, but the differences between categories are not uniform or meaningful. Examples include:
    *   **Education Level:** High School, Bachelor's, Master's, Ph.D.
    *   **Customer Satisfaction:** Very Unsatisfied, Unsatisfied, Neutral, Satisfied, Very Satisfied
    *   **Movie Ratings:** 1-star, 2-star, 3-star, 4-star, 5-star

### 2. Quantitative Data (Numerical Data)

Quantitative data consists of numerical values that represent counts or measurements. This type of data can be measured and expressed numerically, allowing for mathematical operations and statistical analysis.

*   **Discrete Data:** This data consists of distinct, separate values that can be counted. It usually represents whole numbers and cannot be broken down into smaller parts meaningfully. Examples include:
    *   **Number of Children:** 0, 1, 2, 3 (you can't have 2.5 children)
    *   **Number of Cars in a Household:** 1, 2, 3
    *   **Number of Defects on a Product:** 0, 1, 2

*   **Continuous Data:** This data consists of values that can take any value within a given range, including fractions and decimals. It is typically obtained through measurement. Examples include:
    *   **Height:** 175.5 cm, 180.2 cm
    *   **Weight:** 70.3 kg, 68.9 kg
    *   **Temperature:** 25.7°C, 28.1°C
    *   **Time:** 3.14 seconds, 5.87 minutes

### Other Data Classifications

*   **Time Series Data:** Data points collected or recorded at specific time intervals. Often used for trend analysis, forecasting, and anomaly detection. (e.g., stock prices, daily temperature readings).

*   **Cross-Sectional Data:** Data collected by observing many subjects (individuals, firms, countries, etc.) at the same point in time or over a single time period. (e.g., population census, survey of consumer spending in a given month).

*   **Panel Data (Longitudinal Data):** Combines both time series and cross-sectional data, observing the same subjects over multiple time periods. (e.g., tracking the GDP and unemployment rate of several countries over a decade).

Understanding these data types is crucial because it dictates which statistical methods and machine learning algorithms are appropriate for analysis and modeling.

#  Rectangular Data (Data Frames)
## Rectangular data is essentially a two-dimensional matrix with rows indicating records (Observations) and columns indicating features (variables).

### Rectangular Data

The typical frame of reference for an analysis in data science is a rectangular data object, like a spreadsheet or database table.

### KEY TERMS FOR RECTANGULAR DATA

*   **Data frame**
    Rectangular data (like a spreadsheet) is the basic data structure for statistical and machine learning models.

*   **Feature**
    A column in the table is commonly referred to as a feature.

    *   **Synonyms:** attribute, input, predictor, variable

*   **Outcome**
    Many data science projects involve predicting an outcome — often a yes/no outcome (in Table 1-1, it is “auction was competitive or not”). The features are sometimes used to predict the outcome in an experiment or study.

    *   **Synonyms:** dependent variable, response, target, output

*   **Records(Observation)**
    A row in the table is commonly referred to as a record.

    *   **Synonyms:** case, example, instance, observation, pattern, sample

```python
# @title
import pandas as pd
import seaborn as sns

# Load the Titanic dataset
titanic_df = sns.load_dataset('titanic')

# Display the first few rows to show its rectangular structure
print("Titanic Dataset (First 5 Rows):")
display(titanic_df.head())
```

The `titanic_df` we just loaded is an excellent example of **rectangular data**.

Let's break it down using the key terms:

*   **Data frame:** The entire `titanic_df` itself is a **data frame**. It's a two-dimensional, tabular data structure, just like a spreadsheet.

*   **Features:** Each column in the `titanic_df` represents a **feature**. For example:
    *   `survived`: Indicates whether a passenger survived (0 = No, 1 = Yes).
    *   `pclass`: Passenger class (1st, 2nd, or 3rd).
    *   `sex`: Gender of the passenger.
    *   `age`: Age of the passenger.
    *   `fare`: Fare paid by the passenger.
    These features are also known as attributes, inputs, predictors, or variables.

*   **Outcome:** In many analyses involving the Titanic dataset, the `survived` column is often considered the **outcome**. We might try to predict if a passenger survived based on other features like `pclass`, `sex`, `age`, and `fare`. The outcome is also known as the dependent variable, response, target, or output.

*   **Records (Observations):** Each row in the `titanic_df` represents a **record** or an **observation**. For instance, the very first row describes one specific passenger with their `pclass`, `sex`, `age`, `fare`, and `survived` status. Each record provides a complete set of information for a single entity (in this case, a single passenger). These are also known as cases, examples, instances, or samples.

## Non-Rectangular Data

While rectangular data (like the Titanic dataset) is the most common format for statistical and machine learning models, not all data naturally fits into rows and columns. **Non-rectangular data** refers to data structures that are more complex and irregular, often lacking a predefined, fixed schema. This type of data is prevalent in many modern applications, especially with the rise of the internet and diverse data sources.

Key characteristics of non-rectangular data often include:

*   **Variable Schema:** The structure can change from one record to another.
*   **Nested Structures:** Data can contain lists, arrays, or other objects within objects.
*   **Hierarchical Relationships:** Data might be organized in tree-like or graph-like structures.
*   **Unstructured or Semi-structured:** It doesn't always have a strict tabular format.

Examples of non-rectangular data types include:

*   **Text Data:** Raw text documents, social media posts, articles (where each document can have a different length, vocabulary, and internal structure).
*   **Image Data:** Pixel arrays, where dimensions and content vary.
*   **Audio/Video Data:** Time series data of varying lengths and complexities.
*   **Graph Data:** Networks of relationships (e.g., social networks, knowledge graphs).
*   **JSON/XML Data:** Often used for web data, these formats allow for flexible, nested structures.

To use non-rectangular data in traditional rectangular-data-based models, data scientists often perform **feature engineering** to transform it into a tabular format. For instance, with text data, techniques like TF-IDF (Term Frequency-Inverse Document Frequency) or word embeddings are used to convert text into numerical vectors that can then form rows in a data frame.

### Text Data for Spam/Non-Spam Classification

Text data, as we've discussed, is a primary form of non-rectangular data. For spam classification, the raw text of an SMS message or email serves as the input. Each message has a variable length and content, making it unsuitable for direct use in traditional rectangular data models without prior transformation.

The goal of spam classification is to categorize a given message as either 'spam' (unwanted, unsolicited message) or 'ham' (legitimate message). This is a binary classification problem.

```python
# @title
import pandas as pd
from fastai.text.all import *
import requests
import zipfile
import io

# Load the SMS Spam dataset manually from a direct URL (UCI Machine Learning Repository)
print("Loading SMS Spam dataset...")
# The SMS Spam Collection dataset is often found as a tab-separated file
# Corrected URL from UCI ML Repository:
url = 'https://archive.ics.uci.edu/static/public/228/sms+spam+collection.zip'

try:
    # Download the zip file content
    response = requests.get(url)
    response.raise_for_status() # Raise an HTTPError for bad responses (4xx or 5xx)

    # Extract the 'SMSSpamCollection' file from the zip archive
    with zipfile.ZipFile(io.BytesIO(response.content)) as z:
        with z.open('SMSSpamCollection') as f:
            df = pd.read_csv(f, sep='\t', header=None, names=['label', 'text'])

    # Display examples of spam and ham messages
    print("\nExample 'spam' message:")
    print(df[df['label'] == 'spam']['text'].iloc[0])

    print("\nExample 'ham' (non-spam) message:")
    print(df[df['label'] == 'ham']['text'].iloc[0])

    print("\nThis dataset clearly illustrates non-rectangular text data used for a binary classification task: identifying spam.")

except requests.exceptions.RequestException as e:
    print(f"Error downloading or processing the dataset: {e}")
except KeyError:
    print("Error: 'SMSSpamCollection' not found inside the zip file. Check file name.")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
```

### Classification with Fast.ai (SMS Spam Dataset)

Using the Fast.ai SMS Spam dataset, we can clearly see how non-rectangular text data is handled for a **binary classification task**:

*   **Data Type:** The data consists of raw SMS messages, which are variable-length text strings (non-rectangular).

*   **Input (Features):** The text content of each SMS message is the input. To make this usable for a machine learning model, it goes through a similar process as with the IMDB dataset:
    *   **Tokenization:** Breaking messages into words/sub-words.
    *   **Numericalization:** Converting these tokens into numerical IDs.
    *   **Vectorization:** Representing messages as numerical vectors (e.g., using techniques like Bag-of-Words, TF-IDF, or word embeddings).

*   **Output (Outcome/Target for Classification):** The goal is to classify each message into one of two categories: 'spam' or 'ham'.

*   **How it works:** A model learns patterns in the vocabulary, phrasing, and structure of messages associated with either spam or ham. For instance, words like 'free', 'win', 'prize', or specific short codes might be strong indicators of spam, while more conversational language indicates ham. The model then uses these learned distinctions to predict the label of new, incoming messages.

## Introduction to Python

Python is a high-level, interpreted programming language known for its readability and versatility. It's widely used in web development, data analysis, artificial intelligence, scientific computing, and more.

### Key Features of Python:

*   **Simple Syntax:** Python emphasizes readability with clear and concise syntax, making it easier to learn and write code.
*   **Interpreted Language:** Code is executed line by line, which can simplify debugging.
*   **Dynamically Typed:** You don't need to declare the type of a variable when you create it; Python infers it at runtime.
*   **Object-Oriented:** Supports object-oriented programming paradigms.
*   **Extensive Libraries:** A vast ecosystem of libraries and frameworks (like Pandas, NumPy, Matplotlib, Scikit-learn, TensorFlow, PyTorch) makes it powerful for various tasks, especially in data science.
*   **Cross-Platform:** Runs on various operating systems (Windows, macOS, Linux, etc.).

```python
# @title
# 1. Variables and Data Types
# Python supports various data types including numbers (integers, floats), strings, booleans, lists, tuples, dictionaries, and sets.

# Integer
age = 30
print(f"Age: {age}, Type: {type(age)}")

# Float
price = 19.99
print(f"Price: {price}, Type: {type(price)}")

# String
name = "Alice"
print(f"Name: {name}, Type: {type(name)}")

# Boolean
is_active = True
print(f"Is Active: {is_active}, Type: {type(is_active)}")
```

```python
# @title
# 2. Basic Operations
# Arithmetic operations
x = 10
y = 3
print(f"\nAddition: {x + y}")
print(f"Subtraction: {x - y}")
print(f"Multiplication: {x * y}")
print(f"Division: {x / y} (float division)")
print(f"Floor Division: {x // y} (integer division)")
print(f"Modulo: {x % y} (remainder)")
print(f"Exponentiation: {x ** y} (10 to the power of 3)")

# String concatenation
greeting = "Hello, " + name + "!"
print(f"\nGreeting: {greeting}")
```

```python
# @title
# 3. Control Flow (If-Else Statement)
# Conditional statements allow you to execute different code blocks based on conditions.

# Using the 'age' variable defined earlier
if age >= 18:
    print(f"\n{name} is an adult.")
else:
    print(f"\n{name} is a minor.")
```

```python
# @title
# 4. Loops (For Loop)
# Loops are used to iterate over a sequence (like a list, tuple, string, or range).

print("\nCounting from 0 to 4:")
for i in range(5): # range(5) generates numbers 0, 1, 2, 3, 4
    print(i)
```
