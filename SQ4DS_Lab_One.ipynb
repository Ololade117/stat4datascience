# What is Python?

### To a data scientist, Python is the ultimate, non-negotiable tool for turning raw, chaotic data into actionable insights and predictive models. Data scientists see Python as a Swiss Army Knife because it seamlessly bridges the gap between a traditional programming language and a specialized data analytics environment

> **Python is an High-level-Language:** As a high-level language, Python uses English-like syntax, meaning data scientists write code that reads like human logic rather than complex machine instructions.



> **Python is an Object oriented programming Language** Python supports Object-Oriented Programming (OOP) and functional programming. For a data scientist, this means they can bundle data and specific analytical methods together into reusable "objects." They can write a custom data-cleaning function once and easily apply it across millions of data points or entirely different datasets.





> **Python is dynamically typed:** Python is dynamically typed, meaning you do not have to explicitly declare whether a variable is a number, a piece of text, or a massive database table before using it. The language handles this automatically at runtime. This fluidity allows data scientists to rapidly prototype machine learning models and experiment with data variations without getting bogged down by rigid computer science constraints.

> **Python has a huge Library source for data science:** What transforms Python from a general programming language into a data science powerhouse is its open-source package ecosystem. Python hosts thousands of pre-written libraries—like Pandas for data cleaning, Scikit-learn for machine learning, and PyTorch for deep learning. Because these libraries are backed by global tech giants and massive open-source communities, they are continually maintained, optimized, and updated. When new AI breakthroughs or data processing methods emerge, the community rapidly rolls them into existing libraries, ensuring data scientists always have access to cutting-edge tools without rewriting foundational code.

## 15. `while` Loops

A `while` loop repeats while a condition remains true.

````python
count = 1

while count <= 5:
    print(count)
    count += 1
````

A common beginner mistake is creating an infinite loop by forgetting to update the condition.

### `break` and `continue`

`break` exits a loop.

`continue` skips the current iteration.

````python
for number in range(10):
    if number == 5:
        break
    print(number)
````

````python
for number in range(10):
    if number % 2 == 0:
        continue
    print(number)
````

## 17. Understanding `return`

`print()` displays a value.

`return` sends a value back to the code that called the function.

These are different.

````python
def add_and_print(a, b):
    print(a + b)

def add_and_return(a, b):
    return a + b

x = add_and_print(2, 3)
y = add_and_return(2, 3)

print("x =", x)
print("y =", y)
````

## 20. List Comprehensions

A list comprehension is a compact way to build a list.

````python
squares = []

for number in range(1, 6):
    squares.append(number ** 2)

print(squares)
````

The same idea can be written as:

````python
squares = [number ** 2 for number in range(1, 6)]

print(squares)
````

Do not rush to use comprehensions everywhere. First make sure you understand the ordinary loop.

## Problem-Solving Foundations

Before writing code, translate the problem into smaller steps.

For example:

**Problem:** Find the largest number in a list.

### Step 1 — Understand the input

We have a list of numbers.

### Step 2 — Decide what must be tracked

We need a variable containing the largest number seen so far.

### Step 3 — Process the list

Look at each number.

### Step 4 — Update the current answer

If the current number is larger than our current maximum, replace the maximum.

### Step 5 — Return the answer

This way of thinking is more important than memorizing a particular solution.

````python
numbers = [7, 2, 9, 4, 12, 3]

largest = numbers[0]

for number in numbers:
    if number > largest:
        largest = number

print(largest)
````
